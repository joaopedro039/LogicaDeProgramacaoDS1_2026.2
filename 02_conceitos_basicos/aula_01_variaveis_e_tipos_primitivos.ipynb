## 1. O que são Variáveis?
Variáveis são como gavetas etiquetadas na memória do computador. Em Python, a variável é criada automaticamente no momento em que atribuímos um valor a ela com o operador de atribuição `=`.

### Tipos Primitivos:
* **`str` (String):** Textos delimitados por aspas simples `' '` ou duplas `" "`.
* **`int` (Inteiro):** Números inteiros positivos ou negativos (sem casas decimais).
* **`float` (Ponto Flutuante):** Números reais/decimais. Em Python usamos o ponto (`.`) como separador decimal.
* **`bool` (Booleano):** Valores lógicos: `True` (Verdadeiro) ou `False` (Falso).


## 2. Regras de Nomenclatura no Python (PEP 8)
* Use nomes descritivos em letras minúsculas separadas por sublinhado: `snake_case` (ex: `valor_total`, `taxa_servico`).
* **Não pode** começar com números (ex: `1nota` ❌).
* **Não pode** conter espaços (ex: `nome cliente` ❌).
* **Não pode** conter caracteres especiais como acentos ou cedilha (ex: `preço` ❌ -> use `preco` ✅).
* Não use palavras reservadas do Python (como `if`, `for`, `print`, `class`).
